```python
import cv2
import pytesseract
import re

def extract_text(image_path):
    img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
    img = cv2.threshold(img, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)[1]
    
    text = pytesseract.image_to_string(img, config="--psm 6")
    return text.lower()
```

```python
def detect_missing_parts(image_path):
    text = extract_text(image_path)
    missing = []

    for field, patterns in REQUIRED_FIELDS.items():
        found = False
        for pattern in patterns:
            if re.search(pattern, text):
                found = True
                break
        if not found:
            missing.append(field)

    return missing
```

```python
import os

def scan_receipts(folder_path):
    results = {}

    for file in os.listdir(folder_path):
        if file.lower().endswith((".jpg", ".png", ".jpeg")):
            path = os.path.join(folder_path, file)
            missing = detect_missing_parts(path)

            if missing:
                results[file] = {
                    "Label": "INCOMPLETE",
                    "Missing parts": missing
                }
            else:
                results[file] = {
                    "Label": "COMPLETE",
                    "Missing parts": []
                }

    return results
```

```python
results = scan_receipts("receipts")

for img, info in results.items():
    print(f"Image: {img}")
    print("Label:", info["Label"])
    if info["Missing parts"]:
        print("Missing parts:")
        for m in info["Missing parts"]:
            print("-", m)
    else:
        print("No missing parts detected")
    print("-" * 40)
```
