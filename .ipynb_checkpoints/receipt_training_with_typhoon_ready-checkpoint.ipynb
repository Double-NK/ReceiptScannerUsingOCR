```python
import os
import re
import numpy as np
import pandas as pd
import requests
import easyocr

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.model_selection import LeaveOneOut
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
```

```python
USE_TYPHOON_OCR = False

TYPHOON_API_KEY = "PUT_YOUR_API_KEY_HERE"
TYPHOON_API_URL = "PUT_TYHOON_OCR_ENDPOINT_HERE"

ocr_reader = easyocr.Reader(['en'], gpu=False)


def typhoon_ocr(image_path):
    with open(image_path, "rb") as f:
        files = {"file": f}
        headers = {"Authorization": f"Bearer {TYPHOON_API_KEY}"}
        response = requests.post(
            TYPHOON_API_URL,
            files=files,
            headers=headers,
            timeout=60
        )

    if response.status_code != 200:
        raise RuntimeError("Typhoon OCR failed")

    return response.json().get("text", "")


def extract_text(image_path):
    if USE_TYPHOON_OCR:
        try:
            return typhoon_ocr(image_path)
        except Exception:
            pass

    return " ".join(ocr_reader.readtext(image_path, detail=0))
```

```python
def detect_fields(text):
    t = text.lower()

    month_words = (
        "jan|january|feb|february|mar|march|apr|april|may|"
        "jun|june|jul|july|aug|august|sep|september|"
        "oct|october|nov|november|dec|december"
    )

    date_patterns = [
        rf"\d{{1,2}}\s*({month_words})\s*\d{{2,4}}",
        rf"({month_words})\s*\d{{1,2}},?\s*\d{{2,4}}",
        r"\d{1,2}[/-]\d{1,2}[/-]\d{2,4}",
        r"\d{4}[/-]\d{1,2}[/-]\d{1,2}",
    ]

    has_date = any(re.search(p, t) for p in date_patterns) or bool(re.search(r"\b20\d{2}\b", t))

    has_name = any(k in t for k in ["bill to", "billed to", "university", "company", "ltd"])

    has_item_list = any(k in t for k in ["description", "qty", "quantity", "unit price", "total"])

    has_signature = any(k in t for k in ["signature", "received by", "authorized", "cashier", "manager"])

    has_address = any(k in t for k in ["road", "rd", "district", "province", "moo"]) or (
        "chiang rai" in t and re.search(r"\b\d{5}\b", t)
    )

    has_mfu_address = "mae" in t and "fah" in t and "luang" in t

    return {
        "Date": int(has_date),
        "Name": int(has_name),
        "Item list": int(has_item_list),
        "Signature": int(has_signature),
        "Address": int(has_address),
        "MFU address": int(has_mfu_address),
        "Text length": len(t),
        "Digit count": sum(c.isdigit() for c in t)
    }
```

```python
X = []
y = []

def load_dataset(folder, label):
    for img in os.listdir(folder):
        path = os.path.join(folder, img)
        text = extract_text(path)

        if not text.strip():
            continue

        fields = detect_fields(text)

        X.append([
            fields["Date"],
            fields["Name"],
            fields["Item list"],
            fields["Signature"],
            fields["Address"],
            fields["MFU address"],
            fields["Text length"],
            fields["Digit count"]
        ])
        y.append(label)

load_dataset("data/complete", 1)
load_dataset("data/incomplete", 0)

X = np.array(X)
y = np.array(y)

print("Dataset shape:", X.shape)
```

```python
def evaluate_model(model, X, y):
    loo = LeaveOneOut()
    y_true, y_pred = [], []

    for train_idx, test_idx in loo.split(X):
        model.fit(X[train_idx], y[train_idx])
        pred = model.predict(X[test_idx])

        y_true.append(y[test_idx][0])
        y_pred.append(pred[0])

    return {
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred, zero_division=0),
        "Recall": recall_score(y_true, y_pred, zero_division=0),
        "F1": f1_score(y_true, y_pred, zero_division=0),
        "Confusion Matrix": confusion_matrix(y_true, y_pred)
    }
```

```python
lr_model = LogisticRegression(max_iter=1000)
dt_model = DecisionTreeClassifier(max_depth=4, random_state=42)
svm_model = SVC(kernel="linear")

lr_results = evaluate_model(lr_model, X, y)
dt_results = evaluate_model(dt_model, X, y)
svm_results = evaluate_model(svm_model, X, y)

results_df = pd.DataFrame({
    "Logistic Regression": lr_results,
    "Decision Tree": dt_results,
    "SVM": svm_results
})

results_df
```

```python
print("Logistic Regression Confusion Matrix")
print(lr_results["Confusion Matrix"])
print()

print("Decision Tree Confusion Matrix")
print(dt_results["Confusion Matrix"])
print()

print("SVM Confusion Matrix")
print(svm_results["Confusion Matrix"])
```

```python
def find_missing_parts(text):
    fields = detect_fields(text)

    missing = []
    for key in ["Date", "Name", "Item list", "Signature", "Address", "MFU address"]:
        if fields[key] == 0:
            missing.append(key)

    return missing
```

```python
def report_missing(folder):
    for img in os.listdir(folder):
        path = os.path.join(folder, img)
        text = extract_text(path)

        if not text.strip():
            print(f"{img} → Missing: Unable to read text")
            continue

        missing = find_missing_parts(text)
        if missing:
            print(f"{img} → Missing: {', '.join(missing)}")

report_missing("data/complete")
report_missing("data/incomplete")
```
