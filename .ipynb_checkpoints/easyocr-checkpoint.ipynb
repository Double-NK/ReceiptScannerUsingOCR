```python
import os
import re
import cv2
import numpy as np

from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
import easyocr

# Initialize EasyOCR (English receipts)
reader = easyocr.Reader(
    ['en'], 
    gpu=False   # Apple Silicon CPU is fine
)
```

```python
def extract_text(image_path):
    """
    Run OCR and return full text as a single string
    """
    results = reader.readtext(image_path, detail=0)
    text = " ".join(results)
    return text
```

```python
import re

MONTHS = (
    "jan|january|feb|february|mar|march|apr|april|apil|may|"
    "jun|june|jul|july|aug|august|sep|september|"
    "oct|october|nov|november|dec|december"
)

def detect_date(text):
    t = text.lower()

    # normalize OCR whitespace
    t = re.sub(r"\s+", " ", t)

    patterns = [
        # Date: August 17,2024  /  Date: August 17, 2024
        rf"(date|payment date)?\s*:?\s*({MONTHS})\s+\d{{1,2}},?\s*\d{{4}}",

        # Date: 17 August 2024
        rf"(date|payment date)?\s*:?\s*\d{{1,2}}\s+({MONTHS})\s+\d{{4}}",

        # 17/08/2024 or 17-08-2024
        r"\b\d{1,2}[/-]\d{1,2}[/-]\d{2,4}\b",

        # 2024-08-17
        r"\b\d{4}[/-]\d{1,2}[/-]\d{1,2}\b"
    ]

    return any(re.search(p, t) for p in patterns)

def extract_features(text):
    t = text.lower()

    has_date = bool(re.search(rf"\b({MONTHS})|\d{{1,2}}[/-]\d{{1,2}}", t))
    has_name = any(k in t for k in ["co.,", "co.", "company", "ltd", "limited",
        "supply", "shop", "bakery", "restaurant",
        "coffee", "flowers", "service", "center",
        "office", "book", "copy", "printing"])
    has_item = any(k in t for k in ["description", "qty", "amount", "total"])
    has_signature = any(k in t for k in ["signature", "received by", "authorized"])
    has_mfu = "mae fah luang" in t or "mfu" in t

    return [
        int(has_date),
        int(has_name),
        int(has_item),
        int(has_signature),
        int(has_mfu),
        len(t),
        sum(c.isdigit() for c in t)
    ]
```

```python
X = []
y = []

for label, folder in [(1, "data/complete"), (0, "data/incomplete")]:
    for img in os.listdir(folder):
        if img.lower().endswith(".jpg"):
            path = os.path.join(folder, img)
            text = extract_text(path)
            X.append(extract_features(text))
            y.append(label)

X = np.array(X)
y = np.array(y)
```

```python
def analyze_receipt_from_image(image_path):
    text = extract_text(image_path)
    missing = analyze_receipt(text)   # your rule-based function
    return missing
```

```python
def evaluate_model(model, X, y):
    skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

    acc, prec, rec, f1 = [], [], [], []
    cms = []

    for train, test in skf.split(X, y):
        model.fit(X[train], y[train])
        pred = model.predict(X[test])

        acc.append(accuracy_score(y[test], pred))
        prec.append(precision_score(y[test], pred))
        rec.append(recall_score(y[test], pred))
        f1.append(f1_score(y[test], pred))
        cms.append(confusion_matrix(y[test], pred))

    return {
        "Accuracy": np.mean(acc),
        "Precision": np.mean(prec),
        "Recall": np.mean(rec),
        "F1": np.mean(f1),
        "Confusion Matrix": np.sum(cms, axis=0)
    }
```

```python
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Decision Tree": DecisionTreeClassifier(max_depth=5),
    "KNN": KNeighborsClassifier(n_neighbors=5),
    "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42)
}

for name, model in models.items():
    results = evaluate_model(model, X, y)
    print(f"\n{name}")
    for k, v in results.items():
        print(k, ":", v)
```

```python
def detect_vendor_name(text):
    t = text.lower()
    keywords = [
        "company", "co.", "ltd", "limited",
        "service", "bakery", "coffee", "restaurant",
        "supply", "center", "shop", "book", "hotel"
    ]
    return any(k in t for k in keywords)
```

```python
def detect_item_list(text):
    t = text.lower()
    return any(k in t for k in [
        "description", "qty", "quantity",
        "unit price", "amount", "total"
    ])
```

```python
def detect_mfu_address(text):
    t = text.lower()
    return "mae fah luang university" in t or "mfu" in t
```

```python
def detect_signature_ocr(text):
    """
    OCR-based signature detection.
    WARNING: Cannot detect handwritten signatures reliably.
    """
    t = text.lower()
    keywords = [
        "signature",
        "signed by",
        "received by",
        "authorized by",
        "approved by",
        "cashier",
        "manager"
    ]
    return any(k in t for k in keywords)
```

```python
def analyze_receipt(text):
    missing = []

    if not detect_vendor_name(text):
        missing.append("Name")

    if not detect_date(text):     
        missing.append("Date")

    if not detect_item_list(text):
        missing.append("Item list")

    if not detect_mfu_address(text):
        missing.append("MFU address")

    if not detect_signature_ocr(text):
        missing.append("Signature")
    

    return missing
```

```python
def run_all_receipts():
    print("=== COMPLETE RECEIPTS ===")
    for img in sorted(os.listdir("data/complete")):
        if img.lower().endswith(".jpg"):
            path = os.path.join("data/complete", img)
            text = extract_text(path)
            missing = analyze_receipt(text)

            print(f"Receipt: {img}")
            if not missing:
                print("Prediction: COMPLETE")
                print("No missing parts")
            else:
                print("Prediction: INCOMPLETE")
                print("Possible missing field(s):")
                for m in missing:
                    print("-", m)
            print("-" * 40)

    print("\n=== INCOMPLETE RECEIPTS ===")
    for img in sorted(os.listdir("data/incomplete")):
        if img.lower().endswith(".jpg"):
            path = os.path.join("data/incomplete", img)
            text = extract_text(path)
            missing = analyze_receipt(text)

            print(f"Receipt: {img}")
            print("Prediction: INCOMPLETE")
            if missing:
                print("Possible missing field(s):")
                for m in missing:
                    print("-", m)
            else:
                print("No missing parts detected")
            print("-" * 40)
```
